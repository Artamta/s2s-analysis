```python
import cfgrib

path = "/storage/raj.ayush/s2s_final_data/final_iteration/raw/erpas/forecast/annual2023/surface_temperature/tsfc_20230104.grb"
ds = cfgrib.open_dataset(path, backend_kwargs={"indexpath": ""})
da = ds[next(iter(ds.data_vars))]

print("Variable:", da.name)
print("Level type:", da.attrs.get("GRIB_typeOfLevel"))
print("Coordinates:", list(ds.coords))
print("Units:", da.attrs.get("units"))
```

```python
from eccodes import (
    codes_grib_new_from_file,
    codes_get_long,
    codes_get_string,
    codes_release,
)
import pandas as pd

path = (
    "/storage/raj.ayush/s2s_final_data/final_iteration/raw/erpas/"
    "forecast/annual2023/surface_temperature/tsfc_20230104.grb"
)

with open(path, "rb") as file:
    gid = codes_grib_new_from_file(file)

    metadata = {
        "Variable": codes_get_string(gid, "shortName"),
        "Name": codes_get_string(gid, "name"),
        "Units": codes_get_string(gid, "units"),
        "Level type": codes_get_string(gid, "typeOfLevel"),
        "Fixed-surface code": codes_get_long(gid, "typeOfFirstFixedSurface"),
        "Level value": codes_get_long(gid, "level"),
    }

    codes_release(gid)

display(pd.DataFrame(metadata.items(), columns=["GRIB key", "ERPAS value"]))
```
